**RETO 8**

```python

# Reto 8 · Saldo de una estación. Empieza con 12 bicis y capacidad 20.
# Con las salidas y llegadas de 6 horas,
# calcula las bicis al final de cada hora con np.cumsum y np.clip.

# CON VARIABLE ALEATORIA POISSON

bicis = np.random.default_rng(41).poisson(lam = 12, size = (1, 6))

bicis = np.clip(bicis, 0, 20)
print(bicis)

stock = np.cumsum(bicis)
print(stock)

# CON RENTALS Y RETURNS

rentals = np.array([2, 9, 14, 6, 3, 1]) # BICIS ALQUILADAS
returns = np.array([1, 3, 5, 8, 7, 4]) # BICIS DEVUELTAS O APARCADAS

net = returns - rentals
print(12 + np.cumsum(net))

stock = np.clip(12 + np.cumsum(net), 0, 20)
print(stock)
```

**RETO 9**

```python
# 1. np.mean(bikes, axis=1) -> Calcula el promedio de bicicletas por estación.
# Usamos axis=1 para promediar a lo largo de las columnas (las 24 horas) para cada fila (estación).
media_estaciones = np.mean(bikes, axis=1) # Axis = 1 referencia las columnas, es decir, con axis = 1 ignoro el eje 1 que son las columnas, si pusiera axis = 0 ignoro las filas

# 2. np.argmax(bikes, axis=1) -> Obtiene el índice donde se registra el máximo número de bicicletas por fila.
# Como las columnas van de 0 a 23, el índice obtenido se traduce directamente en la hora pico.
hora_max_estaciones = np.argmax(bikes, axis=1)

# 3. Mostramos los resultados iterando ordenadamente sobre los resultados de cada estación
for i in range(len(media_estaciones)):
    print(f"Estación {i}: Media de bicis = {media_estaciones[i]:.2f} | Hora con más bicis = {hora_max_estaciones[i]}h")
```

**RETO 10**

```python
import time

# 1. Creamos una matriz grande de 2000 filas y 500 columnas con números aleatorios
matriz = rng.random((2000, 500))

# --- MÉTODO 1: Usando un bucle clásico de Python ---
inicio_bucle = time.time()
medias_bucle = []
for fila in matriz:
    # Calculamos la media de cada fila de forma manual/básica en Python
    medias_bucle.append(sum(fila) / len(fila))
fin_bucle = time.time()
tiempo_bucle = fin_bucle - inicio_bucle

# --- MÉTODO 2: Usando la vectorización de NumPy ---
inicio_numpy = time.time()
# np.mean con axis=1 calcula la media a lo largo de las columnas para cada fila instantáneamente
medias_numpy = np.mean(matriz, axis=1)
fin_numpy = time.time()
tiempo_numpy = fin_numpy - inicio_numpy

# --- COMPARACIÓN DE RESULTADOS ---
veces_mas_rapido = tiempo_bucle / tiempo_numpy

print(f"Tiempo con bucle clásico: {tiempo_bucle:.6f} segundos")
print(f"Tiempo con NumPy vectorizado: {tiempo_numpy:.6f} segundos")
print(f"\n¡NumPy es {veces_mas_rapido:.2f} veces más rápido que el bucle de Python!")
```
