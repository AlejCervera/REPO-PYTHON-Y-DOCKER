**RETO 7**

```python
# --- RETO 7: Duplicados e imposibles ---
# Guardamos la cantidad de filas iniciales de 'usage'
initial_rows = len(usage)

# Creamos una copia de trabajo de 'usage'
clean = usage.copy()

# Eliminamos las filas que estén completamente duplicadas en el DataFrame
clean = clean.drop_duplicates()
duplicates_removed = initial_rows - len(clean)

# Guardamos la cantidad de registros antes de aplicar el filtro de valores lógicos
rows_before_filter = len(clean)

# Filtramos para quedarnos con registros de alquileres y devoluciones estrictamente entre 0 y 100
clean = clean[
    (clean['rentals'] >= 0) & (clean['rentals'] <= 100) &
    (clean['returns'] >= 0) & (clean['returns'] <= 100)
]
impossible_removed = rows_before_filter - len(clean)

# Imprimimos la contabilidad detallada de los registros eliminados
print(f"Reto 7:")
print(f" - Duplicados eliminados: {duplicates_removed}")
print(f" - Registros con valores imposibles eliminados: {impossible_removed}")
print(f" - Total de filas eliminadas: {initial_rows - len(clean)}")
```

**RETO 8**

```python
# --- RETO 8: Fechas ---
# Convertimos la columna 'timestamp' de string a objeto datetime para poder operar temporalmente
clean['timestamp'] = pd.to_datetime(clean['timestamp'])

# Extraemos únicamente el componente de fecha (año-mes-día)
clean['date'] = clean['timestamp'].dt.date

# Extraemos la hora (formato de 0 a 23)
clean['hour'] = clean['timestamp'].dt.hour

# Obtenemos el número del día de la semana (0 = Lunes, 6 = Domingo)
clean['weekday'] = clean['timestamp'].dt.weekday

# Determinamos si el día de la semana corresponde a fin de semana (Sábado o Domingo)
clean['is_weekend'] = clean['weekday'].isin([5, 6])

print("\nReto 8 completado. Columnas de fecha agregadas.")
print(clean.head())
```

**RETO 9**

```python
# --- RETO 9: Nulos con sentido ---
# Agrupamos por estación (station_id) para no mezclar los datos de disponibilidad entre diferentes puntos de la ciudad.
# Usamos ffill() para propagar hacia adelante el último valor de bicicletas disponibles conocido,
# y bfill() por si los primeros registros de alguna estación fueran nulos.
clean['bikes_available'] = clean.groupby('station_id')['bikes_available'].ffill().bfill()

print("Reto 9 completado. Valores nulos de 'bikes_available' imputados por estación.")
print(clean.isna().sum())
```

**RETO 10**

```python
# --- RETO 10: Distritos y capacidades ---
# Normalizamos el formato de los nombres de los distritos eliminando espacios sobrantes y aplicando formato Título
stations['district'] = stations['district'].str.strip().str.title()

# Imputamos la capacidad nula de cada estación usando la mediana de capacidad calculada para su propio distrito
stations['capacity'] = stations.groupby('district')['capacity'].transform('mean')

print("Reto 10 completado. Nombres de distritos normalizados y capacidades nulas imputadas con la mediana del distrito.\n")

# Mostramos una previsualización rápida de los primeros registros de ambos dataframes limpios
print("Muestra de 'clean':")
display(clean.head())
print("\nMuestra de 'stations':")
display(stations.head())
```
